```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = [1,4,6,7,9]
y = np.pow(x,2)
```

```python
x
```

```python
y
```

Plot Method

```python
plt.plot(x,y)
```

Reduce size of plot using figure

```python
plt.figure(figsize=(3,1)) # width, height
plt.plot(x,y)
```

To make the axis values visible for the ticks that we need

```python
plt.figure(figsize=(8,2)) # width, height
plt.plot(x,y)
plt.xticks([1,2,3,4,5,6,7,8,9])
```

To remove the warnings we use show()

```python
plt.figure(figsize=(8,2)) # width, height
plt.plot(x,y)
plt.xticks([7,5,6,8,9])
plt.yticks([20,50,60,70,120])
plt.show()
```

Add the labels and title for the graph

```python
plt.figure(figsize=(8,2)) # width, height
plt.plot(x,y)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("This is a title")
plt.show()
```

Creating multiple plots in in graph

```python
plt.figure(figsize=(8,4)) # width, height
plt.plot(x,np.pow(x,2)) # square
plt.plot(x,np.log(x)) # log
plt.plot(x,np.pow(x,3)) # cube
plt.show()
```

Giving the legend for the curves - legend() is compusory if the label parameter is used

```python
plt.figure(figsize=(8,4)) # width, height
plt.plot(x,np.pow(x,2), label="square")
plt.plot(x,np.log(x), label="log")
plt.plot(x,np.pow(x,3), label="cube")
plt.legend() ### this is compulsory to represent the legend
plt.show()
```

Subplots -(nrows, ncols, index-cell position)
along with tight_layout()
legends are giving seperately for each box

```python
plt.figure(figsize=(12,6))
plt.subplot(2,3,1) #(nrows, ncols, index)
plt.plot(x,np.pow(x,2), label="square")
plt.legend()
##--------------------------

plt.subplot(2,3,2) #(nrows, ncols, index)
plt.subplot(2,3,3) #(nrows, ncols, index)
plt.plot(x,np.log(x), label="log")
plt.legend()
plt.title("this is log graph")

#------------------
plt.subplot(2,3,4) #(nrows, ncols, index)
plt.subplot(2,3,5) #(nrows, ncols, index)
#-------
plt.subplot(2,3,6) #(nrows, ncols, index)
plt.plot(x,np.pow(x,3), label="cube")
plt.plot(x,np.pow(x,4), label="power4")
plt.legend()
plt.title("title for cube and power4 plot")

#---------------
plt.tight_layout()  ### Adjust the plot spacing and padding
plt.show()
```

Getting grid lines

```python
plt.plot(x,y)
plt.grid()
plt.show()
```

Pie charts

```python
gender = ['male','female','others']
gender_count = [7,5,2]
plt.pie(gender_count,labels=gender,
        autopct='%.2f%%',
        counterclock=False,
        explode=[0,.05,0.1]) ## this is based on order of gender
plt.show()
```
