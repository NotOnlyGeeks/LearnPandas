```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
a1 = np.array([1, 4, 9, 16, 25])
a2 = np.sqrt(a1)
```

```python
plt.figure(figsize=(5,3))
plt.plot(a1, a2)
```

```python
plt.figure(figsize=(5,3))
plt.plot(a1, a2, label="sqrt")
plt.plot(a1, np.log(a1), label="log")
plt.legend()
```

```python
plt.subplot(2,3,1)
plt.plot(a1, a2, label="sqrt")

plt.subplot(2,3,2)

plt.subplot(2,3,3)
plt.plot(a1, np.log(a1), label="log")

plt.subplot(2,3,4)
plt.subplot(2,3,5)
plt.subplot(2,3,6)
plt.tight_layout()
```

```python
plt.subplot(2,3,1)
plt.plot(a1, a2, label="sqrt")
plt.xticks([0,5,10,15,20])
plt.legend()

plt.subplot(2,3,3)
plt.plot(a1, np.log(a1), label="log")
plt.plot(a1, np.exp(a1), label="exp")

plt.legend()

plt.xlabel("x-axis")
plt.ylabel("y-axis")

plt.tight_layout()
```

Seaborn 

```python
import seaborn as sns
```

```python
np.random.seed(0)
gender = np.random.choice(['male','female','others'],size=15)
sns.countplot(x=gender)
```

```python
df = pd.DataFrame({'gender':gender,
                   'age':np.random.randint(1,80,size=15),
                   'place': np.random.choice(['Bengaluru','Mysuru'],size=15)})
df
```

Countplot

```python
plt.figure(figsize=(5,3))
sns.countplot(data=df,x='gender')
plt.show()
```

Add title

```python
plt.figure(figsize=(5,3))
sns.countplot(data=df,x='gender')
plt.title('Count of each gender')
plt.show()
```

Adding bar values

```python
plt.figure(figsize=(5,3))
ax = sns.countplot(data=df,x='gender') ## This is for variable
plt.title('Count of each gender')
ax.bar_label(ax.containers[0],padding=-15)
plt.show()
```

```python
plt.figure(figsize=(10,3))
ax = sns.countplot(data=df,x='gender',hue='place') 
plt.title('Count of each gender')
ax.bar_label(ax.containers[0],padding=-15) ### for gender
ax.bar_label(ax.containers[1],padding=-15) ### for place
plt.show()
```

Getting bar labels for all the bars

```python
plt.figure(figsize=(12,3))
ax = sns.countplot(data=df,x='gender',hue='place') 
plt.title('Count of each gender')

for bar in ax.containers:  # Writing for all the bars in the plot
    ax.bar_label(bar, padding=-15)

plt.show()
```

HistPlot

```python
plt.figure(figsize=(5,3))
sns.histplot(data=df,x='age') ## each bar is a bin
plt.show()
```

Add bar labels

```python
plt.figure(figsize=(5,3))
ax = sns.histplot(data=df,x='age')
ax.bar_label(ax.containers[0],padding=-15) ## for age
plt.show()
```

Giving bins

```python
plt.figure(figsize=(5,3))
ax = sns.histplot(data=df,x='age',bins=7) 
ax.bar_label(ax.containers[0],padding=-15) ## for age
plt.show()
```

```python
plt.figure(figsize=(5,3))
ax = sns.histplot(data=df,x='age',hue='place') 
for bar in ax.containers:  # Writing for values for all the bars in the plot
    ax.bar_label(bar)
plt.show()
```

multiple = ['fill','layer','dodge','stack']

```python
plt.figure(figsize=(5,3))
ax = sns.histplot(data=df,x='age',hue='place',multiple="layer")
for bar in ax.containers:  # Writing for values for all the bars in the plot
    ax.bar_label(bar)
plt.show()
```

```python
plt.figure(figsize=(5,3))
ax = sns.histplot(data=df,x='age',hue='place',multiple="dodge")
for bar in ax.containers:  # Writing for values for all the bars in the plot
    ax.bar_label(bar)
plt.show()
```
